## Question 15:
Given the marks array below - (a) extract all marks greater than 60; (b) extract marks between
50 and 80 using two conditions with & (remember the parentheses!).
marks = np.array([56, 91, 44, 77, 62, 85, 39, 70])

```python
marks = np.array([56, 91, 44, 77, 62, 85, 39, 70])

print("Marks greater than 60:")
print(marks[marks > 60])

print("Marks between 50 and 80:")
print(marks[(marks >= 50) & (marks <= 80)])

```
