## Question 1:


```python
def clean_city(city):
    city = city.strip().title()

    for ch in city:
        if not (ch.isalpha() or ch == " "):
            return "Invalid city"

    return city


city = " nEw dELhi "
print(clean_city(city))
```

## Question 2:

```python
def price_details(prices):
    nums = []

    for price in prices.split(","):
        nums.append(int(price.strip()))

    total = sum(nums)
    highest = max(nums)
    average = total / len(nums)

    print(f"Total : {total}")
    print(f"Highest : {highest}")
    print(f"Average : {average:.2f}")


prices = " 120, 45 , 300 ,80, 95 "
price_details(prices)
```

## Question 3:

```python
def validate_username(username):
    username = username.strip().lower()

    if not username.isalnum():
        return "Invalid username"

    if not username[0].isalpha():
        return "Invalid username"

    if len(username) < 5 or len(username) > 12:
        return "Invalid username"

    return "Valid username"


username = " Rahul123 "
print(validate_username(username))

username = " Rahul_123 "
print(validate_username(username))
```

## Question 4

```python
def sentence_info(sentence):
    words = sentence.split()

    count = len(words)

    vowels = "aeiouAEIOU"
    vowel_end = 0

    for word in words:
        if word[-1] in vowels:
            vowel_end += 1

    longest = max(words, key=len)

    print("Total words :", count)
    print("Words ending with vowel :", vowel_end)
    print("Longest word :", longest)


sentence = "Python is a very powerful language"
sentence_info(sentence)
```

## Question 5:

```python
def char_frequency(text):
    freq = {}

    text = text.replace(" ", "").lower()

    for ch in text:
        freq[ch] = freq.get(ch, 0) + 1

    return freq


text = "Aba c"
print(char_frequency(text))
```

## Question 6:

```python
def word_frequency(sentence):
    freq = {}

    words = sentence.lower().split()

    for word in words:
        freq[word] = freq.get(word, 0) + 1

    print(freq)

    most = max(freq.items(), key=lambda x: x[1])
    print("Most frequent :", most[0])


sentence = "apple mango apple orange mango apple"
word_frequency(sentence)
```

## Question 7:

```python
def invert_sales(sales):
    result = {}

    for day, value in sales.items():
        if value not in result:
            result[value] = []
        result[value].append(day)

    return result


sales = {"mon": 200, "tue": 350, "wed": 200}
print(invert_sales(sales))
```

## Question 8:

```python
def merge_marks(test1, test2):
    result = test1.copy()

    for student, marks in test2.items():
        if student in result:
            result[student] += marks
        else:
            result[student] = marks

    return result


test1 = {"amit": 40, "bina": 35}
test2 = {"amit": 30, "chirag": 45}

print(merge_marks(test1, test2))
```

## Question 9:

```python
def sort_employees(employees):
    employees = sorted(employees, key=lambda x: x[1], reverse=True)

    for name, salary in employees:
        print(f"{name} earns Rs. {salary}")


employees = [("Ravi", 55000), ("Sita", 72000), ("Amit", 48000)]
sort_employees(employees)
```

## Question 10:

```python
def movie_details(movies):
    print("Movies with rating > 8.1")
    for movie in movies:
        if movie["rating"] > 8.1:
            print(movie["title"])

    genres = []

    for movie in movies:
        if movie["genre"] not in genres:
            genres.append(movie["genre"])

    print("\nGenres :", genres)

    ratings = {}

    for movie in movies:
        rating = movie["rating"]

        if rating not in ratings:
            ratings[rating] = []

        ratings[rating].append(movie["title"])

    print("\nMovies grouped by rating")
    for rating in sorted(ratings):
        print(f"{rating} : {ratings[rating]}")


movies = [
    {"title": "Movie A", "rating": 8.2, "genre": "Thriller"},
    {"title": "Movie B", "rating": 8.4, "genre": "Sports"},
    {"title": "Movie C", "rating": 8.1, "genre": "Comedy"},
    {"title": "Movie D", "rating": 8.2, "genre": "Drama"},
]

movie_details(movies)
```
